Importing packages

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Importing and reading data through csv file

```python
df=pd.read_csv("datasets/Data.csv")
df
```

Seperating the dependent and independent variables

```python
X =df.iloc[:,:-1].values
y =df.iloc[:,-1].values
```

```python
X
```

```python
y
```

Fillng the cells with np.nan with mean of the column

```python
from sklearn.impute import SimpleImputer
si=SimpleImputer(missing_values=np.nan,strategy="mean")
si.fit(X[:,1:3])
X[:,1:3]=si.transform(X[:,1:3])
```

```python
X
```

Encoding the Independent catgorical columns

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct=ColumnTransformer(transformers=[("encoder",OneHotEncoder(),[0])],remainder="passthrough")
X=np.array(ct.fit_transform(X))
```

```python
X
```

Encoding the Dependent Catogirical columns

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
y=le.fit_transform(y)
```

```python
y
```

Spiliting into Training and Testing Data

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=1)
```

```python
X_test
```

```python
y_test
```

Feature Scaling through the Standardisation by StandardScaler

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train=sc.fit_transform(X_train)
X_test=sc.transform(X_test)
```
